```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Re-initialize the embedding model
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-mpnet-base-v2"
)

# 2. Check and re-create document splits if they were lost on restart
try:
    all_split
except NameError:
    print("Re-loading and splitting the PDF document...")
    file_path = "https://arxiv.org/pdf/1706.03762"
    loader = PyPDFLoader(file_path)
    doc = loader.load()
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
    all_split = text_splitter.split_documents(doc)

# 3. Connect to vector store and insert documents
vector_store = Chroma(
    collection_name="research_collection",
    embedding_function=embedding_model,
    persist_directory="./chroma_langchain_db"
)

document_ids = vector_store.add_documents(documents=all_split)
print(f"Added {len(document_ids)} documents to the vector store.")

sample = vector_store.get(limit=1, include=["embeddings", "documents"])
print(sample)
```

```python
def retrive_context(query:str,k:int=2):
  retrived_docs=vector.store.similarity_search(query,k=k)
  docs_content=""
  for docs in retrived_docs:
    docs_content+=f"source : {docs.metadata}\n"
    docs_content+=f"content : {docs.page_content}\n"
  return docs_content,retrived_docs
```

```python
def retrive_context(query:str,k:int=2):
  # Fixed the typo from vector.store to vector_store
  retrived_docs=vector_store.similarity_search(query,k=k)
  docs_content=""
  for docs in retrived_docs:
    docs_content+=f"source : {docs.metadata}\n"
    docs_content+=f"content : {docs.page_content}\n"
  return docs_content,retrived_docs

def docu_chat(user_query):
  context,source_docs=retrive_context(user_query,k=2)
  system_message = f"""You are a helpful chatbot.
  Use only the following pieces of context to answer the question.
  Dont try to make up any new information : {context}"""
  message=[
      {"role":"system","content":system_message},
      {"role":"user","content":user_query}
  ]
  # Fixed model.chat to model.invoke
  response=model.invoke(message)
  return {
      "answer":response.content,
      "source_documents":source_docs,
      "context_used":context
  }

result = docu_chat("Explain what is decoder in Transformers")
print(result["answer"])
```
